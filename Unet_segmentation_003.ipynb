```python
# -*- coding: utf-8 -*-
"""U-Net Segmentation From Scratch in Colab"""

# Install necessary library if not present (e.g., for specific augmentations)
# !pip install -q torchinfo albumentations

import os
import numpy as np
import pandas as pd
from PIL import Image
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import torchvision.transforms.functional as TF
from torchvision import transforms
from tqdm.notebook import tqdm # Progress bar
import random
# from torchinfo import summary # Optional: for model summary
# import albumentations as A # Optional: for advanced augmentations
# from albumentations.pytorch import ToTensorV2 # Optional

# --- Configuration ---
# Data Paths (IMPORTANT: Update these paths)
# Option 1: If using Google Drive
# from google.colab import drive
# drive.mount('/content/drive')
# base_dir = '/content/drive/MyDrive/your_segmentation_project/' # Change this

# Option 2: If uploaded directly to Colab session
base_dir = '/content/drive/MyDrive/DATASET' # Example path, create this folder and upload data

TRAIN_IMAGE_DIR = os.path.join(base_dir, '/content/drive/MyDrive/DATASET/s_train_img')
TRAIN_MASK_DIR = os.path.join(base_dir, '/content/drive/MyDrive/DATASET/s_train_mask')
TEST_IMAGE_DIR = os.path.join(base_dir, '/content/drive/MyDrive/DATASET/s_test_img')
TEST_MASK_DIR = os.path.join(base_dir, '/content/drive/MyDrive/DATASET/s_test_mask')

# Create dummy data directories if they don't exist (for testing the script structure)
# In a real scenario, you would upload your data to these paths
if not os.path.exists(base_dir):
    print(f"Creating base directory and dummy data folders at {base_dir}")
    os.makedirs(TRAIN_IMAGE_DIR, exist_ok=True)
    os.makedirs(TRAIN_MASK_DIR, exist_ok=True)
    os.makedirs(TEST_IMAGE_DIR, exist_ok=True)
    os.makedirs(TEST_MASK_DIR, exist_ok=True)
    # Add a few dummy files if needed for testing the loading logic
    # (Pillow can create simple images)
    # Example: Image.new('RGB', (60, 30), color = 'red').save(os.path.join(TRAIN_IMAGE_DIR, 'dummy1.png'))
    # Example: Image.new('L', (60, 30), color = 255).save(os.path.join(TRAIN_MASK_DIR, 'dummy1.png'))
    print("Please upload your actual data to the respective folders.")


# Model & Training Parameters
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {DEVICE}")
LEARNING_RATE = 1e-4
BATCH_SIZE = 8 # Adjust based on GPU memory
NUM_EPOCHS = 25 # Increase for better performance
IMAGE_HEIGHT = 256  # U-Net often requires input size divisible by 16 (or 2^depth)
IMAGE_WIDTH = 256
PIN_MEMORY = True
LOAD_MODEL = False # Set to True to load a saved model
MODEL_SAVE_PATH = os.path.join(base_dir, "unet_segmentation.pth")

# --- U-Net Model Definition ---

class DoubleConv(nn.Module):
    """(Convolution => [BN] => ReLU) * 2"""
    def __init__(self, in_channels, out_channels, mid_channels=None):
        super().__init__()
        if not mid_channels:
            mid_channels = out_channels
        self.double_conv = nn.Sequential(
            nn.Conv2d(in_channels, mid_channels, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(mid_channels),
            nn.ReLU(inplace=True),
            nn.Conv2d(mid_channels, out_channels, kernel_size=3, padding=1, bias=False),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True)
        )

    def forward(self, x):
        return self.double_conv(x)

class Down(nn.Module):
    """Downscaling with MaxPool then DoubleConv"""
    def __init__(self, in_channels, out_channels):
        super().__init__()
        self.maxpool_conv = nn.Sequential(
            nn.MaxPool2d(2),
            DoubleConv(in_channels, out_channels)
        )

    def forward(self, x):
        return self.maxpool_conv(x)

class Up(nn.Module):
    """Upscaling then DoubleConv"""
    def __init__(self, in_channels, out_channels, bilinear=True):
        super().__init__()

        # if bilinear, use the normal convolutions to reduce the number of channels
        if bilinear:
            self.up = nn.Upsample(scale_factor=2, mode='bilinear', align_corners=True)
            self.conv = DoubleConv(in_channels, out_channels, in_channels // 2)
        else:
            # Use Transposed Convolution
            self.up = nn.ConvTranspose2d(in_channels, in_channels // 2, kernel_size=2, stride=2)
            self.conv = DoubleConv(in_channels, out_channels)

    def forward(self, x1, x2):
        # x1 is the output from the previous layer in the decoder
        # x2 is the skip connection from the corresponding layer in the encoder
        x1 = self.up(x1)
        # input is CHW
        diffY = x2.size()[2] - x1.size()[2]
        diffX = x2.size()[3] - x1.size()[3]

        # Pad x1 to match the spatial dimensions of x2 (skip connection)
        x1 = TF.pad(x1, [diffX // 2, diffX - diffX // 2,
                        diffY // 2, diffY - diffY // 2])
        # if you have padding issues, see
        # https://github.com/HaiyongJiang/U-Net-Pytorch-Unstructured-Buggy/commit/0e854509c2cea854e247a9c615f175f76fbb2e3a
        # https://github.com/xiaopeng-liao/Pytorch-UNet/commit/8ebac70e633bac59fc22bb5195e513d5832fb3bd

        # Concatenate along the channel dimension
        x = torch.cat([x2, x1], dim=1)
        return self.conv(x)

class OutConv(nn.Module):
    """Final 1x1 Convolution to map features to output classes"""
    def __init__(self, in_channels, out_channels):
        super(OutConv, self).__init__()
        self.conv = nn.Conv2d(in_channels, out_channels, kernel_size=1)

    def forward(self, x):
        return self.conv(x)


class UNet(nn.Module):
    def __init__(self, n_channels, n_classes, bilinear=True, base_features=64):
        super(UNet, self).__init__()
        self.n_channels = n_channels
        self.n_classes = n_classes
        self.bilinear = bilinear
        factor = 2 if bilinear else 1 # Factor for channel reduction in Up block if bilinear

        # Encoder (Contracting Path)
        self.inc = DoubleConv(n_channels, base_features)
        self.down1 = Down(base_features, base_features * 2)
        self.down2 = Down(base_features * 2, base_features * 4)
        self.down3 = Down(base_features * 4, base_features * 8)
        self.down4 = Down(base_features * 8, base_features * 16 // factor) # Adjusted for bilinear

        # Decoder (Expansive Path)
        self.up1 = Up(base_features * 16, base_features * 8 // factor, bilinear)
        self.up2 = Up(base_features * 8, base_features * 4 // factor, bilinear)
        self.up3 = Up(base_features * 4, base_features * 2 // factor, bilinear)
        self.up4 = Up(base_features * 2, base_features, bilinear)

        # Final Output Layer
        self.outc = OutConv(base_features, n_classes)

    def forward(self, x):
        x1 = self.inc(x)
        x2 = self.down1(x1)
        x3 = self.down2(x2)
        x4 = self.down3(x3)
        x5 = self.down4(x4) # Bottleneck output

        # Decoder with skip connections
        x = self.up1(x5, x4)
        x = self.up2(x, x3)
        x = self.up3(x, x2)
        x = self.up4(x, x1)
        logits = self.outc(x)
        return logits

# --- Dataset Definition ---
class SegmentationDataset(Dataset):
    def __init__(self, image_dir, mask_dir, transform=None, target_transform=None):
        self.image_dir = image_dir
        self.mask_dir = mask_dir
        self.transform = transform
        self.target_transform = target_transform # Separate transform for masks

        # Ensure filenames match (handle potential extensions like _mask.png)
        self.images = sorted([f for f in os.listdir(image_dir) if not f.startswith('.')]) # Ignore hidden files
        self.masks = sorted([f for f in os.listdir(mask_dir) if not f.startswith('.')])

        # Basic check: Make sure we have image-mask pairs
        # This assumes simple matching names. Adapt if your naming is complex.
        if len(self.images) != len(self.masks):
             print(f"Warning: Mismatch in number of images ({len(self.images)}) and masks ({len(self.masks)}). Trying to match...")
             # Simple matching logic: image 'a.png' matches mask 'a.png' or 'a_mask.png'
             img_stems = {os.path.splitext(img)[0] for img in self.images}
             mask_stems = {os.path.splitext(msk)[0].replace('_mask', '').replace('_segmentation','') for msk in self.masks} # common suffixes
             common_stems = list(img_stems.intersection(mask_stems))

             self.images = sorted([img for img in self.images if os.path.splitext(img)[0] in common_stems])
             # Need to find the corresponding mask name for each common stem
             self.masks = []
             mask_files = os.listdir(mask_dir)
             for stem in common_stems:
                 found = False
                 potential_masks = [f"{stem}.png", f"{stem}.jpg", f"{stem}.tif", f"{stem}_mask.png", f"{stem}_segmentation.png"] # Add expected formats/suffixes
                 for potential_mask in potential_masks:
                     if potential_mask in mask_files:
                          self.masks.append(potential_mask)
                          found = True
                          break
                 if not found:
                     print(f"Could not find mask for image stem: {stem}")

             self.masks.sort() # Keep them sorted after matching

             if len(self.images) != len(self.masks) or len(self.images) == 0:
                raise ValueError("Could not resolve image/mask pairs. Check filenames and directories.")
             print(f"Successfully matched {len(self.images)} image/mask pairs.")


    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):
        img_path = os.path.join(self.image_dir, self.images[idx])
        mask_path = os.path.join(self.mask_dir, self.masks[idx])

        # Load image and mask
        image = Image.open(img_path).convert("RGB") # Ensure 3 channels
        mask = Image.open(mask_path).convert("L")   # Ensure grayscale (single channel)

        # --- Apply Transforms ---
        # Use a synchronized transform approach if doing spatial augmentations (like rotation, flip)
        # Seed for synchronization if using random transforms
        seed = np.random.randint(2147483647) # make a seed with numpy generator

        # Apply image transforms
        if self.transform:
            random.seed(seed) # PyTorch transforms use Python's random module
            torch.manual_seed(seed) # Set torch seed if using torch.*random transforms
            image = self.transform(image)

        # Apply mask transforms (often simpler: resize, tensor, maybe threshold)
        if self.target_transform:
            random.seed(seed)
            torch.manual_seed(seed)
            mask = self.target_transform(mask)

            # Ensure mask is binary (0 or 1) - Important for BCE loss
            # Assuming mask pixels are 0 (background) and 255 (object)
            mask = (mask > 0.5).float() # Threshold after converting to tensor [0,1] range

        return image, mask

# --- Transformations ---
# Define transforms (adjust normalization based on your dataset)
# Note: Using simple transforms here. Consider Albumentations for more complex augmentations.

# For Training Data (with augmentation)
train_img_transform = transforms.Compose([
    transforms.Resize((IMAGE_HEIGHT, IMAGE_WIDTH)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomVerticalFlip(p=0.1),
    # transforms.RandomRotation(10), # Optional
    # transforms.ColorJitter(brightness=0.2, contrast=0.2), # Optional
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]), # ImageNet stats
])

train_mask_transform = transforms.Compose([
    transforms.Resize((IMAGE_HEIGHT, IMAGE_WIDTH)),
    transforms.RandomHorizontalFlip(p=0.5), # Must match image flip
    transforms.RandomVerticalFlip(p=0.1),  # Must match image flip
    # transforms.RandomRotation(10), # Optional, must match image rotation
    transforms.ToTensor(),
])


# For Validation/Test Data (no augmentation, just resize and normalize)
val_img_transform = transforms.Compose([
    transforms.Resize((IMAGE_HEIGHT, IMAGE_WIDTH)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
])

val_mask_transform = transforms.Compose([
    transforms.Resize((IMAGE_HEIGHT, IMAGE_WIDTH)),
    transforms.ToTensor(),
])


# --- Dice Coefficient Metric ---
def dice_coefficient(preds, targets, smooth=1e-6):
    """Computes the Dice coefficient."""
    preds = torch.sigmoid(preds) # Apply sigmoid to get probabilities [0,1]
    preds = (preds > 0.5).float() # Threshold to get binary mask {0,1}

    # Flatten label and prediction tensors
    preds = preds.view(-1)
    targets = targets.view(-1)

    intersection = (preds * targets).sum()
    dice = (2. * intersection + smooth) / (preds.sum() + targets.sum() + smooth)
    return dice

# --- Training Function ---
def train_fn(loader, model, optimizer, loss_fn, scaler=None): # Added scaler for mixed precision
    loop = tqdm(loader, desc="Training", leave=True)
    running_loss = 0.0
    running_dice = 0.0

    model.train() # Set model to training mode

    for batch_idx, (data, targets) in enumerate(loop):
        data = data.to(device=DEVICE)
        # Ensure target is FloatTensor and correct shape [Batch, 1, H, W]
        targets = targets.float().to(device=DEVICE)

        # Forward pass
        if DEVICE == "cuda" and scaler is not None:
             # Mixed precision training
            with torch.cuda.amp.autocast():
                predictions = model(data)
                loss = loss_fn(predictions, targets)
                dice = dice_coefficient(predictions, targets)
        else:
            # Standard precision training
            predictions = model(data)
            loss = loss_fn(predictions, targets)
            dice = dice_coefficient(predictions, targets)


        # Backward pass and optimization
        optimizer.zero_grad()
        if DEVICE == "cuda" and scaler is not None:
            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()
        else:
            loss.backward()
            optimizer.step()

        # Update statistics
        running_loss += loss.item()
        running_dice += dice.item()

        # Update progress bar
        loop.set_postfix(loss=loss.item(), dice=dice.item())

    avg_loss = running_loss / len(loader)
    avg_dice = running_dice / len(loader)
    return avg_loss, avg_dice

# --- Evaluation Function ---
def check_accuracy(loader, model, loss_fn, device="cuda"):
    num_correct = 0
    num_pixels = 0
    dice_score = 0
    total_loss = 0.0
    model.eval() # Set model to evaluation mode

    with torch.no_grad(): # No need to track gradients during evaluation
        loop = tqdm(loader, desc="Evaluating", leave=False)
        for x, y in loop:
            x = x.to(device)
            y = y.float().to(device) # Ensure target is float

            preds = model(x)
            loss = loss_fn(preds, y)
            total_loss += loss.item()

            preds_prob = torch.sigmoid(preds)
            preds_binary = (preds_prob > 0.5).float()

            # Pixel Accuracy (Optional, Dice is usually more relevant for segmentation)
            num_correct += (preds_binary == y).sum().item()
            num_pixels += torch.numel(preds_binary) # Total number of pixels

            # Dice Score
            dice_score += dice_coefficient(preds, y).item() # Use raw logits for dice func

            loop.set_postfix(loss=loss.item(), dice=dice_coefficient(preds, y).item())


    avg_loss = total_loss / len(loader)
    avg_dice = dice_score / len(loader)
    pixel_accuracy = (num_correct / num_pixels) * 100 if num_pixels > 0 else 0

    print(f"Evaluation Results:")
    print(f"  Average Loss: {avg_loss:.4f}")
    print(f"  Dice Score: {avg_dice:.4f}")
    print(f"  Pixel Accuracy: {pixel_accuracy:.2f}%")

    model.train() # Set back to training mode if needed later
    return avg_loss, avg_dice, pixel_accuracy


# --- Prediction Visualization ---
def show_predictions(loader, model, num_samples=5, device="cuda"):
    model.eval()
    images_shown = 0
    fig, axes = plt.subplots(num_samples, 3, figsize=(10, num_samples * 4))
    fig.suptitle("Sample Predictions vs Ground Truth", fontsize=16)
    axes[0, 0].set_title("Input Image")
    axes[0, 1].set_title("Ground Truth Mask")
    axes[0, 2].set_title("Predicted Mask")


    with torch.no_grad():
        for x, y in loader:
            x = x.to(device)
            preds = model(x)
            preds = torch.sigmoid(preds)
            preds = (preds > 0.5).float() # Apply threshold

            # Move data to CPU for plotting & revert normalization for visualization
            x_cpu = x.cpu()
            y_cpu = y.cpu().squeeze(1) # Remove channel dim for plotting
            preds_cpu = preds.cpu().squeeze(1) # Remove channel dim

            # Inverse normalization (approximate for visualization)
            inv_normalize = transforms.Normalize(
               mean=[-0.485/0.229, -0.456/0.224, -0.406/0.225],
               std=[1/0.229, 1/0.224, 1/0.225]
            )

            for i in range(x_cpu.shape[0]): # Iterate through batch
                 if images_shown >= num_samples:
                     break

                 img_display = inv_normalize(x_cpu[i]).permute(1, 2, 0) # C,H,W -> H,W,C
                 img_display = torch.clamp(img_display, 0, 1) # Clamp values for display

                 ax_row = axes[images_shown]
                 ax_row[0].imshow(img_display)
                 ax_row[0].axis('off')

                 ax_row[1].imshow(y_cpu[i], cmap='gray')
                 ax_row[1].axis('off')

                 ax_row[2].imshow(preds_cpu[i], cmap='gray')
                 ax_row[2].axis('off')

                 images_shown += 1

            if images_shown >= num_samples:
                break

    plt.tight_layout(rect=[0, 0.03, 1, 0.97]) # Adjust layout to prevent title overlap
    plt.show()
    model.train() # Set back to train mode

# --- Main Execution ---
def main():
    print("Setting up datasets and dataloaders...")
    # Create Datasets
    train_dataset = SegmentationDataset(
        image_dir=TRAIN_IMAGE_DIR,
        mask_dir=TRAIN_MASK_DIR,
        transform=train_img_transform,
        target_transform=train_mask_transform
    )
    test_dataset = SegmentationDataset(
        image_dir=TEST_IMAGE_DIR,
        mask_dir=TEST_MASK_DIR,
        transform=val_img_transform,
        target_transform=val_mask_transform
    )

    # Check if datasets are empty
    if len(train_dataset) == 0:
        print("ERROR: Training dataset is empty. Please check the TRAIN_IMAGE_DIR and TRAIN_MASK_DIR paths and ensure data exists.")
        return
    if len(test_dataset) == 0:
        print("ERROR: Test dataset is empty. Please check the TEST_IMAGE_DIR and TEST_MASK_DIR paths and ensure data exists.")
        return

    # Create DataLoaders
    train_loader = DataLoader(
        train_dataset,
        batch_size=BATCH_SIZE,
        shuffle=True,
        num_workers=os.cpu_count() // 2, # Adjust based on Colab resources
        pin_memory=PIN_MEMORY,
    )
    test_loader = DataLoader(
        test_dataset,
        batch_size=BATCH_SIZE, # Can often use larger batch size for testing
        shuffle=False,
        num_workers=os.cpu_count() // 2,
        pin_memory=PIN_MEMORY,
    )
    print(f"Found {len(train_dataset)} training images, {len(test_dataset)} test images.")
    print(f"Train loader batches: {len(train_loader)}, Test loader batches: {len(test_loader)}")


    print("Initializing model...")
    # Initialize Model (3 input channels for RGB, 1 output class for binary segmentation)
    model = UNet(n_channels=3, n_classes=1, base_features=64).to(DEVICE)

    # Optional: Print model summary
    # summary(model, input_size=(BATCH_SIZE, 3, IMAGE_HEIGHT, IMAGE_WIDTH))

    # Loss Function (Binary Cross Entropy with Logits is numerically stable)
    # It combines Sigmoid layer and BCELoss
    loss_fn = nn.BCEWithLogitsLoss()

    # Optimizer
    optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE)

    # Learning Rate Scheduler (Optional but recommended)
    # scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, 'max', patience=2, factor=0.1, verbose=True) # Monitor Dice score

    # Gradient Scaler for Mixed Precision (if using GPU)
    scaler = torch.cuda.amp.GradScaler() if DEVICE == "cuda" else None


    # Load model checkpoint if specified
    if LOAD_MODEL and os.path.exists(MODEL_SAVE_PATH):
        print(f"Loading model from {MODEL_SAVE_PATH}")
        checkpoint = torch.load(MODEL_SAVE_PATH, map_location=DEVICE)
        model.load_state_dict(checkpoint['state_dict'])
        # optimizer.load_state_dict(checkpoint['optimizer']) # Optional: load optimizer state
        print("Model loaded.")


    # --- Training Loop ---
    print("Starting training...")
    best_dice = -1.0
    train_losses, train_dices = [], []
    val_losses, val_dices = [], []

    for epoch in range(NUM_EPOCHS):
        print(f"\n--- Epoch {epoch+1}/{NUM_EPOCHS} ---")

        # Train one epoch
        avg_train_loss, avg_train_dice = train_fn(train_loader, model, optimizer, loss_fn, scaler)
        train_losses.append(avg_train_loss)
        train_dices.append(avg_train_dice)
        print(f"Epoch {epoch+1} Train Avg Loss: {avg_train_loss:.4f}, Train Avg Dice: {avg_train_dice:.4f}")


        # Evaluate on test set
        avg_val_loss, avg_val_dice, _ = check_accuracy(test_loader, model, loss_fn, device=DEVICE)
        val_losses.append(avg_val_loss)
        val_dices.append(avg_val_dice)

        # Optional: Adjust learning rate based on validation Dice
        # scheduler.step(avg_val_dice)

        # Save the model if it has the best validation dice score so far
        if avg_val_dice > best_dice:
            print(f"Validation Dice improved ({best_dice:.4f} --> {avg_val_dice:.4f}). Saving model...")
            best_dice = avg_val_dice
            checkpoint = {
                "state_dict": model.state_dict(),
                "optimizer": optimizer.state_dict(), # Save optimizer state too
                "epoch": epoch,
                "best_dice": best_dice,
            }
            torch.save(checkpoint, MODEL_SAVE_PATH)
            print(f"Model saved to {MODEL_SAVE_PATH}")

    print("\nTraining finished.")

    # --- Plot Training Curves ---
    print("Plotting training curves...")
    plt.figure(figsize=(12, 5))
    plt.subplot(1, 2, 1)
    plt.plot(range(1, NUM_EPOCHS + 1), train_losses, label='Train Loss')
    plt.plot(range(1, NUM_EPOCHS + 1), val_losses, label='Validation Loss')
    plt.title('Loss over Epochs')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()

    plt.subplot(1, 2, 2)
    plt.plot(range(1, NUM_EPOCHS + 1), train_dices, label='Train Dice')
    plt.plot(range(1, NUM_EPOCHS + 1), val_dices, label='Validation Dice')
    plt.title('Dice Coefficient over Epochs')
    plt.xlabel('Epoch')
    plt.ylabel('Dice Coefficient')
    plt.legend()
    plt.tight_layout()
    plt.show()


    # --- Final Evaluation & Predictions ---
    print("\nPerforming final evaluation on the test set using the best model...")
    # Load the best model saved during training
    if os.path.exists(MODEL_SAVE_PATH):
         print(f"Loading best model from {MODEL_SAVE_PATH} for final evaluation.")
         checkpoint = torch.load(MODEL_SAVE_PATH, map_location=DEVICE)
         model.load_state_dict(checkpoint['state_dict'])
    else:
         print("Warning: No saved model found. Evaluating with the model from the last epoch.")

    check_accuracy(test_loader, model, loss_fn, device=DEVICE)

    print("\nShowing some predictions on the test set...")
    show_predictions(test_loader, model, num_samples=5, device=DEVICE) # Show 5 sample predictions

    print("\nScript finished.")


if __name__ == "__main__":
    # Create dummy data folders if they don't exist (for initial run)
     if not os.path.exists(base_dir):
        os.makedirs(TRAIN_IMAGE_DIR, exist_ok=True)
        os.makedirs(TRAIN_MASK_DIR, exist_ok=True)
        os.makedirs(TEST_IMAGE_DIR, exist_ok=True)
        os.makedirs(TEST_MASK_DIR, exist_ok=True)
        print(f"Created dummy data folders in {base_dir}. Please upload your data.")
        # Optionally add placeholder files to avoid errors during dry run
        # (comment out if data exists)
        # try:
        #     Image.new('RGB', (60, 30), color = 'red').save(os.path.join(TRAIN_IMAGE_DIR, 'dummy1.png'))
        #     Image.new('L', (60, 30), color = 255).save(os.path.join(TRAIN_MASK_DIR, 'dummy1.png'))
        #     Image.new('RGB', (60, 30), color = 'blue').save(os.path.join(TEST_IMAGE_DIR, 'dummy_test1.png'))
        #     Image.new('L', (60, 30), color = 0).save(os.path.join(TEST_MASK_DIR, 'dummy_test1.png'))
        # except Exception as e:
        #     print(f"Could not create dummy files: {e}")

     main()
```
